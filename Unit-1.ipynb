### 3. Write a program to perform arithmetic relational and logical operations using Python operators

```python
num1=int(input("enter 1st number:"))
num2=int(input("enter 2nd number:"))
num3=int(input("enter 3rd number:"))

print("\n-------Arithmetic operations-------")
print(f"sum of {num1} and {num2} is ",num1+num2)
print(f"sub of {num1} and {num2} is ",num1-num2)
print(f"mul of {num1} and {num2} is ",num1*num2)
print(f"div of {num1} and {num2} is ",num1/num2)

print("\n-------Relational operations-------")
print(f"{num1} is greater than {num2} =",num1>num2)
print(f"{num2} is greater than {num1} =",num1<num2)

print("\n-------Logical operations-------")
print(f"{num1} is greater than {num2} and {num1} is greater than {num3} = {num1 > num2 and num1 > num3}")
print(f"{num1} is greater than {num2} or {num1} is greater than {num3} = {num1 < num2 or num1 > num3}")
```

### 4. Write a program to demonstrate string operations including slicing formatting and built-in string functions. 

```python
str=input("Enter string here")
age=input("Enter age here")

print("------String slicing and fromatting------")
print(str[0:3])
print(str[0:6:2])
print(str[::-1])
print(f"My name is {str} my age is {age}")

print("------String in-built functions------")
print(str.upper())
print(str.find('a'))
print(str.join(age))
print(str.count('a'))
print(str.replace('a','4'))
print(str.split(' '))
```

### 9. Write a program to define and use user-defined functions with different types of arguments. 

```python
def display():
    print("Inside the display function")
display()

print("function with (positional argument)")
def add(num1,num2):
    return num1+num2
addition=add(10,20)
print('Addition:',addition)

print("function with (keyword argument)")
name=input('Enter your name:')
age=input("Enter age:")
def user_info(name,age):
    print('Name:',name)
    print('Age:',age)
user_info(name,age)
user_info(name,age=20)

print("function with (variable length argument)")
def emp_info(*name,dept):
    print("name:",name)
    print("department:",dept)

emp_info("pranav","harsh",dept="sales")
```

### 10.Write a program to demonstrate recursion using factorial or Fibonacci series. 

```python
n=int(input("Enter number to get "))

def fact(n):
    if(n<0):
        return
    if(n<2):
        return 1
    return n*fact(n-1)
factorial=fact(n)
print(factorial)

def fibonacci(n):
   if n <= 1:
       return n
   else:
       return fibonacci(n - 1) + fibonacci(n - 2)

terms = 9
for i in range(terms):
   print(fibonacci(i), end=" ")
```
