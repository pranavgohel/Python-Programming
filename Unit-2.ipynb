### 2. Write a program to check whether a number is positive negative or zero using nested conditions.

```python
n1=int(input("enter any number:"))

if n1!=0:
    if n1>0:
        print("The number is positive")
    elif n1<0:
        print("The number is negative")
else:
    print("The number is 0")
```

### 3. Write a program to generate a multiplication table using a for loop.

```python
n=int(input("Enter table no:-"))

print(f"Table no {n}")
for i in range(1,11):
    print(f"{n}*{i}=",n*i)
```

### 4. Write a program to find the sum of digits of a number using a while loop. 

```python
n=int(input("enter any number:"))

digit_sum=0
while n>0:
    digit=n%10
    digit_sum+=digit
    n =n//10
print(digit_sum)
```

### 8. Write a program to illustrate variable scope using local global and nonlocal variables. 

```python
g_var="global variable"

def display():
    l_var="local variable"
    print("inside function")
    print("printing local variable",l_var)
display()
print("outside the function")
print("printing global variable",g_var)

def outer():
    count=0

    def inner():
        nonlocal count  # Refers to 'count' in outer()
        count+=1
        print("Inner count:",count)

    inner()
    inner()
    print("Outer count:",count)

outer()
```
